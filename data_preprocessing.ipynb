```python
from dgp.datasets import SynchronizedSceneDataset

dataset = SynchronizedSceneDataset(
    'ddad.json',
    datum_names=('lidar', 'camera_01'),
    generate_depth_from_datum='lidar',
    split='train'
)
```

```python
import os
import imageio
import numpy as np
from tqdm import tqdm

# 저장할 디렉터리 생성
os.makedirs('output/rgb', exist_ok=True)
os.makedirs('output/depth_gt_sparse', exist_ok=True)

# 데이터셋 순회
for idx, sample in enumerate(tqdm(dataset)):
    cam = sample[0][0]  # CAMERA_01 데이터 (OrderedDict)

    # RGB 이미지 저장
    rgb = cam['rgb']  # PIL.Image 객체
    rgb.save(f'output/rgb/{idx:06d}.png')

    # Depth map (float32, meters)
    depth = cam['depth']  # numpy array, shape: (H, W)

    # 보간 없이: 유효한 픽셀만 포함 (depth > 0), 나머지는 0 유지
    depth_mm = (depth * 1000.0).astype(np.uint16)  # meters → millimeters

    # PNG로 저장 (0은 무효 픽셀로 유지됨)
    imageio.imwrite(f'output/depth_gt_sparse/{idx:06d}.png', depth_mm)
```

```python
# 데이터셋 구조를 확인하는 코드
for idx, sample in enumerate(tqdm(dataset)):
    print(sample)  # sample[0], sample[1] 각각의 데이터 구조를 출력하여 확인
    break  # 첫 번째 샘플만 출력
```
